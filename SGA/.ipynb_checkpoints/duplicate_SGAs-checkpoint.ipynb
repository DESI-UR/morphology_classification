# Investigating Duplicates in the SGA Catalogs

Results so far:

10 true duplicates in SGA 2025 (2 are blank sky)

113 closely interacting galaxies discovered

```python
import os
import glob
import matplotlib.pyplot as plt
from PIL import Image
from collections import defaultdict
```

```python
# Directory catalog
sga2020 = "/pscratch/sd/q/qshimp/Cutouts/sga2020/Coord_matches"
sga2025 = "/pscratch/sd/q/qshimp/Cutouts/sga2025/Coord_matches"
sgatest = "/pscratch/sd/q/qshimp/Cutouts/sga2020"
```

```python
def get_png_groups(folder, roundRA=2, roundDec=2):

    files = glob.glob(os.path.join(folder, "*.png"))

    groups = defaultdict(list)
    coords = {}

    # First pass: exact rounded matches
    for f in files:

        name = os.path.basename(f).replace(".png", "")
        parts = name.split("_")

        try:
            ra = float(parts[-2])
            dec = float(parts[-1])

            coords[f] = (ra, dec)

            key = (round(ra, roundRA),
                   round(dec, roundDec))

            groups[key].append(f)

        except Exception:
            continue

    duplicate_groups = []
    leftovers = []

    for g in groups.values():
        if len(g) > 1:
            duplicate_groups.append(g)
        else:
            leftovers.extend(g)

    print(f"Round-match groups: {len(duplicate_groups)}")
    print(f"Leftover images: {len(leftovers)}")

    # ----------------------------------------
    # Second pass: near-rounding matches
    # ----------------------------------------

    paired = set()
    near_groups = []

    step_ra = 10**(-roundRA)
    step_dec = 10**(-roundDec)

    for i, f1 in enumerate(leftovers):

        if f1 in paired:
            continue

        ra1, dec1 = coords[f1]

        group = [f1]

        for f2 in leftovers[i+1:]:

            if f2 in paired:
                continue

            ra2, dec2 = coords[f2]

            if (abs(ra1 - ra2) <= step_ra and
                abs(dec1 - dec2) <= step_dec):

                group.append(f2)
                paired.add(f2)

        if len(group) > 1:
            paired.add(f1)
            near_groups.append(group)

    unpaired = [f for f in leftovers if f not in paired]

    print(f"Near-match groups: {len(near_groups)}")
    print(f"Still unpaired: {len(unpaired)}")

    return duplicate_groups, near_groups, unpaired
    

def display_dups(duplicate_groups):
    # Display pairs
    for group in duplicate_groups:
    
        fig, axes = plt.subplots(
            1,
            len(group),
            figsize=(5 * len(group), 5)
        )
    
        # Make sure axes is always iterable
        if len(group) == 1:
            axes = [axes]
        else:
            axes = axes.flatten()
    
        for ax, file in zip(axes, group):
    
            img = Image.open(file)
    
            ax.imshow(img)

            # Get image dimensions
            width, height = img.size

            # Small X at the exact center
            ax.plot(
                width / 2,
                height / 2,
                marker="x",
                markersize=8,
                markeredgewidth=2
            )

            ax.set_title(os.path.basename(file), fontsize=8)
            ax.axis("off")
    
        plt.tight_layout()
        plt.show()
```

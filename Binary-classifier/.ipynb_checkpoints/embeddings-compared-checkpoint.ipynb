### Anchor Embeddings
Use: generate embeddings from h5 files generated because they are not present in SGA 2025. 
Instructions: change RUN_NAME to match that of the desired generated file

```python
from ssl_legacysurvey.utils import plotting_tools as plt_tools
from ssl_legacysurvey.utils import load_data, format_logger
from ssl_legacysurvey.data_loaders import datamodules
from ssl_legacysurvey.moco.moco2_module import Moco_v2

from dataclasses import dataclass, field
from pathlib import Path
import time
import torch
import pickle
import numpy as np
from tqdm.notebook import tqdm
```

```python
@dataclass
class Config:
    output_dir: Path = Path("/pscratch/sd/q/qshimp/Sorter")
    ssl_dir: str = Path("/global/cfs/cdirs/desicollab/users/ioannis/SGA/2025/ssl")

    seed: int = 67
    checkpoint: Path = Path("/global/homes/q/qshimp/ssl-legacysurvey/resnet50.ckpt")

    npix: int = 152
    embedding_batch_size: int = 128

    augmentations: str = "rrjc"
    jitter_lim: int = 0

# Change these
RUN_NAME = "JJ_emb_comparison"
VERBOSE = 2

config = Config()
anchor_h5_path = str("global/cfs/cdirs/desicollab/users/ioannis/SGA/2025/ssl/ssl-cutouts-dr11-south-chunk0007.hdf5")

run_path = config.output_dir / "binary_classifier" / RUN_NAME
run_path.mkdir(parents=True, exist_ok=True)
```

```python
def build_ssl_params(config):
    return {
        "augmentations": config.augmentations,
        "npix_out": config.npix,
        "jitter_lim": config.jitter_lim,
        "verbose": False,
    }
 
 
def process_images(dataset, processed_images_path, verbose=1):
    processed_images_path = Path(processed_images_path)
    if processed_images_path.exists():
        if verbose > 1:
            print(f"Loaded processed images from {processed_images_path}")
        t0 = time.perf_counter()
        images = torch.load(processed_images_path, map_location="cpu")
        if verbose > 1:
            print(f"  torch.load took {time.perf_counter() - t0:.1f}s")
        return images
 
    sample_image, _ = dataset[0]
    images = torch.empty((len(dataset), *sample_image.shape), dtype=sample_image.dtype)
 
    iterator = dataset
    if verbose > 1:
        iterator = tqdm(dataset, desc="Processing images")
    else:
        print(f"Processing {len(dataset)} images (heartbeat every 5000, set verbose>1 for a live bar)...")
 
    t_start = time.perf_counter()
    for i, (image, _) in enumerate(iterator):
        images[i] = image
        if i > 0 and i % 5000 == 0:
            elapsed = time.perf_counter() - t_start
            rate = i / elapsed
            eta_min = (len(dataset) - i) / rate / 60
            print(f"  {i}/{len(dataset)} images | {rate:.1f} img/s | ETA {eta_min:.1f} min")
 
    torch.save(images, processed_images_path)
    if verbose:
        print(f"Saved processed images to {processed_images_path} "
              f"(took {time.perf_counter() - t_start:.1f}s total)")
    return images
 
 
def generate_embeddings(images, backbone, config, output_path, verbose):
    if output_path is not None and output_path.exists():
        if verbose > 1:
            print(f"Loaded embeddings from {output_path}")
        return torch.load(output_path)
 
    if len(images) == 0:
        raise ValueError("Cannot generate embeddings from empty image set.")
 
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    backbone = backbone.to(device)
    backbone.eval()
 
    embeddings = []
    with torch.no_grad():
        for start in tqdm(
            range(0, len(images), config.embedding_batch_size),
            desc="Generating embeddings",
            disable=(verbose < 2),
        ):
            end = min(start + config.embedding_batch_size, len(images))
            batch = images[start:end].to(device)
            batch_embeddings = backbone(batch).cpu()
            embeddings.append(batch_embeddings)
 
    embeddings = torch.cat(embeddings)
    if output_path is not None:
        torch.save(embeddings, output_path)
        if verbose:
            print(f"Saved embeddings to {output_path}")
 
    return embeddings
```

```python
# ---- Anchors: still need real image -> embedding generation ----
loader = load_data.DecalsDataLoader(image_dir=anchor_h5_path, npix_in=config.npix)
 
anchor_gals = loader.get_data(-1, fields=["sgaid", "ra", "dec", "Main_Type"], npix_out=config.npix)
anchor_indices = np.where(anchor_gals)
 
model = Moco_v2.load_from_checkpoint(checkpoint_path=config.checkpoint)
backbone = model.encoder_q
backbone.fc = torch.nn.Identity()
 
params = build_ssl_params(config)
transform = datamodules.DecalsTransforms(config.augmentations, {"npix_out": config.npix, "jitter_lim": config.jitter_lim})
 
# Build the (index, dummy_label) lookup file DecalsDataset actually expects
label_array = np.stack([anchor_indices, np.zeros_like(anchor_indices)], axis=1).astype(np.int64)
label_path = run_path / "anchor_label_index.npy"
np.save(label_path, label_array)
 
dataset = datamodules.DecalsDataset(anchor_h5_path, str(label_path), transform, params)
 
processed_images_path = run_path / "anchor_images.pt"
anchor_images = process_images(dataset, processed_images_path, VERBOSE)
 
anchor_embeddings = generate_embeddings(anchor_images, backbone, config, run_path / "anchor_embeddings.pt", VERBOSE)
  
bridge_path = run_path / "anchor_artifacts.pkl"
bridge_data = {
    "anchor_gals": anchor_gals,
    "anchor_embeddings": anchor_embeddings.detach().cpu().numpy(),
}
with open(bridge_path, "wb") as f:
    pickle.dump(bridge_data, f)
print(f"Saved anchor artifacts to {bridge_path}")
```

```python
import pickle
from pathlib import Path
from dataclasses import dataclass, field
import numpy as np
from astropy.table import vstack
from astropy.coordinates import SkyCoord
import astropy.units as u
from SGA.ssl import load_ssl_embeddings

# ============================================================
# CONFIGURATION -- must match notebook 1's RUN_NAME/output_dir
# so this reads the bridge artifact that notebook 1 just wrote.
# ============================================================
RUN_NAME = "sga2025_sample_test3"
OUTPUT_DIR = Path("/pscratch/sd/q/qshimp/Sorter")
SSL_DIR = "/global/cfs/cdirs/desicollab/users/ioannis/SGA/2025/ssl"
UNKNOWN_TYPE = 30
DEDUP_RADIUS_ARCSEC = 2.0  # verify against the separation histogram before trusting this

run_path = OUTPUT_DIR / "binary_classifier" / RUN_NAME


# ============================================================
# Load notebook 1's output
# ============================================================
def load_anchor_bridge(run_path):
    bridge_path = run_path / "anchor_artifacts.pkl"
    with open(bridge_path, "rb") as f:
        bridge_data = pickle.load(f)
    print(f"Loaded {len(bridge_data['anchor_gals']['targetid'])} anchors from {bridge_path}")
    return bridge_data["anchor_gals"], bridge_data["anchor_embeddings"]


anchor_gals, anchor_embeddings = load_anchor_bridge(run_path)


# ============================================================
# Load the main SGA2025 catalog (already has embeddings -- no
# ssl_legacysurvey / torch needed for this population)
# ============================================================
def load_combined_ssl_data(ssl_dir='/global/cfs/cdirs/desicollab/users/ioannis/SGA/2025/ssl'):
    south = load_ssl_embeddings('dr11-south', ssl_dir)
    north = load_ssl_embeddings('dr11-north', ssl_dir)
    south['region'] = 'dr11-south'   # tag before combining, so downstream
    north['region'] = 'dr11-north'   # region-specific lookups (cutout_index,
    return vstack([south, north])    # viewer links) stay correct per-row

main_data = load_combined_ssl_data()
print(f"Loaded {len(main_data):,d} main-catalog galaxies")


# ============================================================
# Harmonize both sources to a common schema (no mags -- dropped per earlier decision)
# ============================================================
def harmonize_anchor_gals(gals):
    return {
        "targetid": np.asarray(gals["targetid"]),
        "ra": np.asarray(gals["ra"]),
        "dec": np.asarray(gals["dec"]),
        "Main_Type": np.asarray(gals["Main_Type"]),
    }


def harmonize_main_catalog(data, unknown_type=UNKNOWN_TYPE):
    n = len(data)
    return {
        "targetid": np.asarray(data["SGAID"]),
        "ra": np.asarray(data["RA"]),
        "dec": np.asarray(data["DEC"]),
        "Main_Type": np.full(n, unknown_type, dtype=np.int32),
    }


anchor_common = harmonize_anchor_gals(anchor_gals)
main_common = harmonize_main_catalog(main_data)
main_embeddings = np.asarray(main_data["embeddings"])


# ============================================================
# Sky-match dedup: drop main-catalog rows that are actually anchors,
# so the anchor's real VI label is the only copy retained
# ============================================================
def dedup_main_catalog_against_anchors(anchor_common, main_common, main_embeddings,
                                        radius_arcsec=DEDUP_RADIUS_ARCSEC, verbose=True):
    anchor_coords = SkyCoord(ra=anchor_common["ra"] * u.deg, dec=anchor_common["dec"] * u.deg)
    main_coords = SkyCoord(ra=main_common["ra"] * u.deg, dec=main_common["dec"] * u.deg)

    idx, sep2d, _ = anchor_coords.match_to_catalog_sky(main_coords)
    is_duplicate = sep2d < (radius_arcsec * u.arcsec)

    dup_main_indices = idx[is_duplicate]
    keep_mask = np.ones(len(main_common["ra"]), dtype=bool)
    keep_mask[dup_main_indices] = False

    if verbose:
        print(f"Sky-matched {is_duplicate.sum()} / {len(anchor_common['ra'])} anchors "
              f"to main-catalog rows within {radius_arcsec}\" -- "
              f"removing those duplicates ({keep_mask.sum()} / {len(keep_mask)} main-catalog rows kept)")

    main_common_deduped = {key: val[keep_mask] for key, val in main_common.items()}
    main_embeddings_deduped = main_embeddings[keep_mask]
    return main_common_deduped, main_embeddings_deduped


main_common, main_embeddings = dedup_main_catalog_against_anchors(
    anchor_common, main_common, main_embeddings
)


# ============================================================
# Merge into one population
# ============================================================
def merge_gals_and_embeddings(anchor_common, anchor_embeddings, main_common, main_embeddings):
    merged_gals = {
        key: np.concatenate([anchor_common[key], main_common[key]])
        for key in anchor_common
    }
    merged_embeddings = np.concatenate([anchor_embeddings, main_embeddings], axis=0)
    return merged_gals, merged_embeddings


merged_gals, merged_embeddings = merge_gals_and_embeddings(
    anchor_common, anchor_embeddings, main_common, main_embeddings
)
print(f"Merged population: {len(merged_gals['ra']):,d} galaxies "
      f"({len(anchor_common['ra']):,d} anchors + {len(main_common['ra']):,d} main-catalog)")


# ============================================================
# STOP HERE -- compute_umap needs umap-learn, which lives in the
# ssl-pl kernel. Save this bridge for notebook 3 (binary_classifier)
# to load and pick up from compute_umap onward.
# ============================================================
bridge_path = run_path / "merged_artifacts.pkl"
with open(bridge_path, "wb") as f:
    pickle.dump({"merged_gals": merged_gals, "merged_embeddings": merged_embeddings}, f)
print(f"Saved merged artifacts to {bridge_path}")
print("Notebook 3 picks up from here: load this file, then run compute_umap(merged_embeddings, ...) onward.")
```

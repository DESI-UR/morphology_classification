```python
# Standard library
import hashlib
import json
import os
import pickle
from pathlib import Path
from dataclasses import dataclass, field
from typing import Optional, Tuple
from tqdm import tqdm
import time

# Scientific Python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

# SSL Legacy Survey
from ssl_legacysurvey.utils import plotting_tools as plt_tools

# Machine learning
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score, classification_report
from sklearn.model_selection import train_test_split, KFold, cross_val_score, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Dimensionality reduction
from umap import UMAP
```

```python
def inverse_squared_distance(distances):
    distances = np.maximum(distances, 1e-10)
    return distances ** -2

@dataclass
class ClassifierResults:
    class_id: int
    class_name: str
    pipeline: Pipeline
    metrics: dict
    parameters: dict
    training_summary: dict
    prediction_catalogue: pd.DataFrame

@dataclass
class Config:
    """
    Runtime configuration for the binary classifier.
    """
    output_dir: Path
    seed: int = 67
    npix: int = 152
    unknown_type: int = 30

    # -------------------------------------------------
    # KNN
    # -------------------------------------------------
    knn_neighbors: int = 10
    knn_weights: str = "distance"
    knn_metric: str = "cosine"
    knn_cv: int = 5
    tune_knn: bool = False
    knn_param_grid: tuple = (
        {
            "n_neighbors": [3, 5, 7, 10, 25, 50, 75, 90, 100],
            "weights": ["uniform", "distance", inverse_squared_distance],
            "algorithm": ["ball_tree", "kd_tree"],
            "metric": ["minkowski", "euclidean", "manhattan", "chebyshev"]
        },
        {
            "n_neighbors": [3, 5, 7, 10, 25, 50, 75, 90, 100],
            "weights": ["uniform", "distance", inverse_squared_distance],
            "algorithm": ["brute"],
            "metric": ["minkowski", "euclidean", "manhattan", "chebyshev", "cosine"]
        }
    )

    # -------------------------------------------------
    # Plotting
    # -------------------------------------------------
    plot_types: Tuple[str, ...] = ("CMs", "Overlays")
    scatter_nx: int = 40
    scatter_ny: int = 60

    # -------------------------------------------------
    # Column mappings
    # -------------------------------------------------
    galaxy_types: dict = field(default_factory=lambda: {
        -5: {"name": "Irregular",   "color": "#E69F00"},  # orange
         0: {"name": "Lenticular",  "color": "#56B4E9"},  # sky blue
        10: {"name": "Spiral",      "color": "#009E73"},  # green
        20: {"name": "Elliptical",  "color": "#CC79A7"},  # magenta
        30: {"name": "Unknown",     "color": "#999999"}   # gray
    })

    target_id_column: str = "targetid"
    embedding_column: str = "Embeddings"
    umap_column: str = "UMAP_Coordinates"
    main_type_column: str = "Main_Type"
    galaxy_type_column: str = "Galaxy_Type"
    binary_label_column: str = "Binary_Label"
    prediction_column: str = "Predicted_Type"

    @property
    def morphology_names(self):
        return {
            key: value["name"]
            for key, value in self.galaxy_types.items()
        }
```

```python
def prepare_output_directory(config, name): # Create output directory structure    
    run_path = config.output_dir / "binary_classifier" / name
    run_path.mkdir(parents=True, exist_ok=True)
    return run_path

def get_classes(run_type, config):  # Return morphology class IDs    
    run_type = run_type.strip().lower()

    if run_type == "all":
        return [
            class_id
            for class_id in config.galaxy_types.keys()
            if class_id != config.unknown_type
        ]

    for class_id, info in config.galaxy_types.items():
        if info["name"].lower() == run_type:
            if class_id == config.unknown_type:
                raise ValueError("Unknown is not a valid classifier target.")
            return [class_id]

    valid = ", ".join(
        ["all"] + [
            info["name"].lower()
            for class_id, info in config.galaxy_types.items()
            if class_id != config.unknown_type
        ]
    )
    raise ValueError(f"Unknown run_type '{run_type}'. Valid options are: {valid}")

def load_merged_artifacts(bridge_run_path):
    artifacts = np.load(bridge_run_path / "merged_artifacts.npz", allow_pickle=True)
    gals = {
        "targetid": artifacts["targetid"],
        "ra": artifacts["ra"],
        "dec": artifacts["dec"],
        "Main_Type": artifacts["Main_Type"],
    }
    embeddings = artifacts["embeddings"]
    umap_embeddings = artifacts["umap_embeddings"]
    return gals, embeddings, umap_embeddings

def build_dataframe(gals, embeddings, umap_embeddings, config):
    embeddings = embeddings.detach().cpu().numpy() if hasattr(embeddings, "detach") else np.asarray(embeddings)
    df = pd.DataFrame({
        config.target_id_column: gals["targetid"],
        "target_ra": gals["ra"],
        "target_dec": gals["dec"],
        config.embedding_column: list(embeddings),
        config.umap_column: list(umap_embeddings),
        config.main_type_column: gals["Main_Type"],
    })
    df[config.galaxy_type_column] = df[config.main_type_column].map(lambda x: get_type_name(x, config))
    return df
```

```python
def make_plot_paths(run_path, iteration, name, config, class_name=None):
    iteration_path = run_path / "iterations" / f"{iteration:02d}"
    paths = {}

    for plot_type in config.plot_types:
        folder = iteration_path / "plots" / plot_type
        folder.mkdir(parents=True, exist_ok=True)

        if plot_type in ("CMs", "Overlays") and class_name is not None:
            filename = f"{name}_{class_name.lower()}_it_{iteration:02d}.png"
        else:
            filename = f"{name}_it_{iteration:02d}.png"

        paths[plot_type] = folder / filename

    return paths

def get_type_name(class_id, config): # Retrieve morphology
    return config.galaxy_types[class_id]["name"]

def get_type_color(class_id, config): # Retrieve color for plotting
    return config.galaxy_types[class_id]["color"]

def split_known_unknown(df, config): # Split anchors from unclassified set
    known = df[df[config.main_type_column] != config.unknown_type].copy()
    unknown = df[df[config.main_type_column] == config.unknown_type].copy()
    return known, unknown

def build_binary_labels(known_df, class_id, config): # Use main type to label "True" class
    known_df = known_df.copy()
    known_df[config.binary_label_column] = (known_df[config.main_type_column] == class_id).astype(int)
    return known_df
    
def tune_knn(X_train, y_train, config, verbose=1):
    if verbose > 2:
        print("Starting KNN hyperparameter tuning...")
    grid_search = GridSearchCV(estimator=KNeighborsClassifier(), param_grid=config.knn_param_grid, cv=config.knn_cv)
    grid_search.fit(X_train, y_train)
    if verbose > 2:
        print("\nTuning complete!\nBest Parameters:")
        print(grid_search.best_params_)
        print("\nBest Estimator:")
        print(grid_search.best_estimator_)
        print(f"Best cross-validation accuracy: {grid_search.best_score_:.3f}")
    return grid_search
```

```python
def train_binary_classifier(train_df, knn, config):
    X_train = np.vstack(train_df[config.embedding_column])
    y_train = train_df[config.binary_label_column].values
            
    pipeline = Pipeline([("scaler", StandardScaler()), ("knn", knn)])
    pipeline.fit(X_train, y_train)

    return pipeline

def evaluate_classifier(pipeline, test_df, config, verbose=1):
    X_test = np.vstack(test_df[config.embedding_column])
    y_test = test_df[config.binary_label_column].values
    y_pred = pipeline.predict(X_test)

    if verbose > 2:
        print(classification_report(y_test, y_pred))

    report = classification_report(y_test, y_pred, output_dict=True)
    return {"report": report, "y_test": y_test.tolist(), "y_pred": y_pred.tolist()}

def predict_dataframe(pipeline, dataframe, class_id, config):
    prediction_df = dataframe.copy()
    prediction_df[config.prediction_column] = (prediction_df[config.main_type_column])
    unknown_mask = (prediction_df[config.main_type_column] == config.unknown_type)

    if not unknown_mask.any():
        return prediction_df

    print("Building unknown matrix...")
    t0 = time.perf_counter()

    X_unknown = np.vstack(
        prediction_df.loc[unknown_mask, config.embedding_column]
    )

    print(f"Stack took {time.perf_counter()-t0:.1f}s")

    print(f"Predicting {len(X_unknown):,} galaxies...")
    t0 = time.perf_counter()

    predictions = pipeline.predict(X_unknown)

    print(f"Prediction took {time.perf_counter()-t0:.1f}s")
    
    prediction_df.loc[unknown_mask, config.prediction_column] = np.where(predictions == 1, class_id, config.unknown_type)

    return prediction_df

def classify_one_type(class_id, dataframe, config, run_path, iteration, plots=True, verbose=1):
    class_name = config.galaxy_types[class_id]["name"]
    class_key = class_name.lower()
    
    cache_path = (run_path / "classifier_cache" / f"it_{iteration:02d}_{class_key}.pkl")
    cache_path.parent.mkdir(parents=True, exist_ok=True)
    
    if cache_path.exists():
        if verbose:
            print(f"Loaded cached {class_name} classifier.")
        with open(cache_path, "rb") as f:
            return pickle.load(f)
    
    t0 = time.perf_counter()
    known_df, unknown_df = split_known_unknown(dataframe, config)
    print(f"  [timing] split_known_unknown: {time.perf_counter()-t0:.1f}s ({len(known_df)} known, {len(unknown_df)} unknown)")

    t0 = time.perf_counter()
    known_df = known_df.sample(frac=1, random_state=config.seed).reset_index(drop=True)
    known_df = build_binary_labels(known_df, class_id, config)
    print(f"  [timing] sample+build_binary_labels: {time.perf_counter()-t0:.1f}s")

    t0 = time.perf_counter()
    train_df, test_df = train_test_split(
        known_df, test_size=0.2, random_state=config.seed,
        stratify=known_df[config.binary_label_column],
    )
    print(f"  [timing] train_test_split: {time.perf_counter()-t0:.1f}s")

    if config.tune_knn:
        t0 = time.perf_counter()
        X_train = np.vstack(train_df[config.embedding_column])
        y_train = train_df[config.binary_label_column].values
        tuning = tune_knn(X_train, y_train, config, verbose)
        knn = tuning.best_estimator_
        print(f"  [timing] tune_knn: {time.perf_counter()-t0:.1f}s")
    else:
        knn = KNeighborsClassifier(
            n_neighbors=config.knn_neighbors,
            weights=config.knn_weights,
            metric=config.knn_metric,
        )

    t0 = time.perf_counter()
    pipeline = train_binary_classifier(train_df, knn, config)
    print(f"  [timing] train_binary_classifier: {time.perf_counter()-t0:.1f}s")

    t0 = time.perf_counter()
    metrics = evaluate_classifier(pipeline, test_df, config, verbose)
    print(f"  [timing] evaluate_classifier: {time.perf_counter()-t0:.1f}s")

    t0 = time.perf_counter()
    predictions = predict_dataframe(pipeline, dataframe, class_id, config)
    print(f"  [timing] predict_dataframe: {time.perf_counter()-t0:.1f}s ({(dataframe[config.main_type_column]==config.unknown_type).sum()} unknown rows predicted)")

    training_summary = {
        "train_size": len(train_df),
        "test_size": len(test_df),
        "positive": int(train_df[config.binary_label_column].sum()),
        "negative": int(len(train_df) - train_df[config.binary_label_column].sum()),
    }
    results = ClassifierResults(
        class_id=class_id,
        class_name=class_name,
        pipeline=pipeline,
        prediction_catalogue=predictions,
        metrics=metrics,
        training_summary=training_summary,
        parameters=(
            tuning.best_params_
            if config.tune_knn
            else {
                "n_neighbors": config.knn_neighbors,
                "weights": config.knn_weights,
                "metric": config.knn_metric,
            }
        ),
    )
    
    with open(cache_path, "wb") as f:
        pickle.dump(results, f)
    
    if verbose:
        print(f"Cached {class_name} classifier.")
    
    return results
```

```python
def save_CM(y_true, y_pred, class_name, output_path, verbose=1):
    if output_path.exists():
        if verbose > 1:
            print(f"Loaded existing plot: {output_path}")
        return
        
    labels = {0: "Other", 1: class_name}
    cm = confusion_matrix(y_true, y_pred, normalize="true")

    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=list(labels.values()))
    fig, ax = plt.subplots(figsize=(6, 6))
    disp.plot(cmap="Blues", values_format=".2f", ax=ax, colorbar=False)

    ax.set_title(f"{class_name} Binary Classifier")
    ax.set_xlabel("Predicted")
    ax.set_ylabel("True")

    fig.tight_layout()
    fig.savefig(output_path, dpi=300, bbox_inches="tight")
    if verbose > 1:
        print(f"Saved confusion matrix: {output_path}")
    if verbose > 2:
        print(f"Accuracy: {accuracy_score(y_true, y_pred):.3f}")
        plt.show()
    plt.close(fig)

def save_overlay(prediction_df, class_id, output_path, config, verbose=1):
    if output_path.exists():
        if verbose > 1:
            print(f"Loaded existing plot: {output_path}")
        return

    all_coords = np.vstack(prediction_df[config.umap_column])

    plt.figure(figsize=(10, 8))
    plt.scatter(all_coords[:, 0], all_coords[:, 1], color="lightgray", s=3, alpha=0.35, label="All galaxies")

    predicted = prediction_df[
        (prediction_df[config.main_type_column] == config.unknown_type)
        &
        (prediction_df[config.prediction_column] == class_id)
    ]

    if len(predicted) > 0:
        coords = np.vstack(predicted[config.umap_column])
        plt.scatter(
            coords[:, 0],
            coords[:, 1],
            color=get_type_color(class_id, config),
            s=12,
            label=get_type_name(class_id, config),
        )

    plt.title(f"{get_type_name(class_id, config)} Classifier Predictions")
    plt.xlabel("UMAP Dimension 1")
    plt.ylabel("UMAP Dimension 2")
    plt.legend()
    plt.tight_layout()
    plt.savefig(output_path, dpi=300, bbox_inches="tight")

    if verbose > 1:
        plt.show()
    plt.close()
    if verbose:
        print(f"New {get_type_name(class_id, config)} classifications: {len(predicted)}")

def file_exists(path, verbose=1):
    path = Path(path)
    if path.exists():
        if verbose > 1:
            print(f"Using existing file: {path}")
        return True
    return False

def save_output(results, iteration, run_path, name, config, plots=True, verbose=1):
    prediction_dir = run_path / "classifications"
    prediction_dir.mkdir(parents=True, exist_ok=True)
    info_dir = run_path / "info"
    info_dir.mkdir(parents=True, exist_ok=True)

    prediction_file = prediction_dir / f"{iteration:02d}_{results.class_name.lower()}_predictions.csv"
    if not file_exists(prediction_file, verbose):
        results.prediction_catalogue.to_csv(prediction_file, index=False)

    summary = {
        "class_name": results.class_name,
        "class_id": results.class_id,
        **results.training_summary,
        "parameters": results.parameters,
        "metrics": results.metrics,
    }

    summary_file = (info_dir / f"{iteration:02d}_{results.class_name.lower()}_summary.json")

    if not file_exists(summary_file, verbose):
        with open(summary_file, "w") as f:
            json.dump(summary, f, indent=4)

    if plots:
        plot_paths = make_plot_paths(run_path, iteration, name, config, class_name=results.class_name)
        save_CM(results.metrics["y_test"], results.metrics["y_pred"], results.class_name, plot_paths["CMs"], verbose)
        save_overlay(results.prediction_catalogue, results.class_id, plot_paths["Overlays"], config, verbose)

    model_dir = run_path / "models"
    model_dir.mkdir(parents=True, exist_ok=True)   
    model_file = (model_dir / f"{iteration:02d}_{results.class_name.lower()}_pipeline.pkl")
    
    if not file_exists(model_file, verbose):   
        with open(model_file, "wb") as f:
            pickle.dump(results.pipeline, f)

    if verbose:
        print(f"Saved {results.class_name} classifier")
        if verbose > 2:
            print(f"  Predictions : {prediction_file}")
            print(f"  Summary     : {summary_file}")

def combine_results(binary_results, config, return_stats=False):
    if not binary_results:
        raise ValueError("No classifier results supplied.")

    first_result = next(iter(binary_results.values()))
    combined = first_result.prediction_catalogue[[config.target_id_column, config.main_type_column]].copy()
    prediction_columns = []

    for result in binary_results.values():
        column = f"Prediction_{result.class_name}"
        combined[column] = (result.prediction_catalogue[config.prediction_column].values)
        prediction_columns.append(column)

    unknown_mask = (combined[config.main_type_column] == config.unknown_type)
    predictions = (combined.loc[unknown_mask, prediction_columns].to_numpy())

    valid = predictions != config.unknown_type
    agreements = valid.sum(axis=1)
    stats = pd.Series(agreements).value_counts().sort_index()
    accepted = agreements == 1

    if not np.any(accepted):
        result = pd.DataFrame(columns=[config.target_id_column, config.prediction_column])
        if return_stats:
            return result, stats
        return result

    accepted_predictions = []

    for row in np.where(accepted)[0]:
        accepted_predictions.append(predictions[row][valid[row]][0])
    result = pd.DataFrame(
        {
            config.target_id_column:
                combined.loc[unknown_mask].iloc[np.where(accepted)[0]][config.target_id_column].values,
            config.prediction_column:
                accepted_predictions,
        }
    )
    if return_stats:
        return result, stats
    return result

def update_catalog(df, targets, config):
    updated = df.copy()

    if len(targets) == 0:
        return updated

    prediction_map = dict(zip(targets[config.target_id_column], targets[config.prediction_column]))

    updated[config.main_type_column] = (
        updated[config.target_id_column]
        .map(prediction_map)
        .fillna(updated[config.main_type_column])
        .astype(int)
    )
    updated[config.galaxy_type_column] = (updated[config.main_type_column].map(lambda x: get_type_name(x, config)))
    
    return updated
```

```python
def evaluate_accuracy_by_size(data, config, size_column="D26", n_bins=4, verbose=True):
    """
    Bin galaxies by angular size (size_column) and evaluate KNN classification
    accuracy within each bin, using the same KNN settings as the main
    classifier (config.knn_neighbors, knn_weights, knn_metric).

    Only galaxies with a known morphology label (excludes config.unknown_type)
    are used, since accuracy requires ground truth to score against.

    Returns a DataFrame: one row per size bin, with n_galaxies, median D26,
    and held-out accuracy.
    """
    label_col = config.main_type_column
    known = data[data[label_col] != config.unknown_type]
    df = known.to_pandas() if hasattr(known, "to_pandas") else pd.DataFrame(known)

    df["size_bin"] = pd.qcut(df[size_column], q=n_bins, duplicates="drop")

    results = []
    for bin_label, group in df.groupby("size_bin", observed=True):
        if len(group) < config.knn_cv * 2:
            if verbose:
                print(f"Skipping bin {bin_label} -- only {len(group)} galaxies, too few")
            continue

        X = np.stack(group["embeddings"].to_numpy())
        y = group[label_col].to_numpy()

        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.3, random_state=config.seed, stratify=y
        )

        pipeline = Pipeline([
            ("scaler", StandardScaler()),
            ("knn", KNeighborsClassifier(
                n_neighbors=config.knn_neighbors,
                weights=config.knn_weights,
                metric=config.knn_metric,
            )),
        ])
        pipeline.fit(X_train, y_train)
        acc = pipeline.score(X_test, y_test)

        results.append({
            "size_bin": str(bin_label),
            "n_galaxies": len(group),
            "median_D26": group[size_column].median(),
            "accuracy": acc,
        })
        if verbose:
            print(f"{bin_label}: n={len(group):>6d}  "
                  f"median_D26={group[size_column].median():>6.2f}  accuracy={acc:.3f}")

    return pd.DataFrame(results)
```

```python
def save_agreement_histogram(stats, output_path, verbose):
    if output_path.exists():
        return
    plt.figure(figsize=(7,5))
    plt.bar(stats.index.astype(str), stats.values)
    plt.xlabel("Number of types claimed")
    plt.ylabel("Galaxy count")
    plt.title("Binary Classifier Agreement")
    plt.tight_layout()
    plt.savefig(output_path,dpi=300)
    if verbose > 1:
        plt.show()
    plt.close()

def save_iteration_overlay(df, added, config, output_path, verbose):
    if output_path.exists():
        return

    coords = np.vstack(df[config.umap_column])

    plt.figure(figsize=(10, 8))

    # Background: all galaxies
    plt.scatter(
        coords[:, 0], coords[:, 1],
        s=3,
        color="lightgray",
        alpha=0.3,
        label="All galaxies"
    )

    # Build lookup from targetid -> predicted class
    prediction_map = dict(zip(
        added[config.target_id_column],
        added[config.prediction_column]
    ))
    
    for class_id, info in config.galaxy_types.items():
        if class_id == config.unknown_type:
            continue
    
        # Which targetids were predicted as this class?
        ids = [
            tid for tid, pred in prediction_map.items()
            if pred == class_id
        ]
    
        if not ids:
            continue
    
        mask = df[config.target_id_column].isin(ids)
        added_coords = np.vstack(df.loc[mask, config.umap_column])
    
        plt.scatter(
            added_coords[:, 0],
            added_coords[:, 1],
            s=15,
            color=info["color"],
            label=info["name"],
        )

    plt.legend()
    plt.title("New Classifications Added")
    plt.tight_layout()
    plt.savefig(output_path, dpi=300)

    if verbose > 1:
        plt.show()

    plt.close()

def save_iteration(df_updated, iteration, run_path, config, info=True, verbose=1):
    run_path.mkdir(parents=True, exist_ok=True)
    checkpoint = run_path / f"iteration_{iteration:02d}.pkl"

    with open(checkpoint, "wb") as f:
        pickle.dump(df_updated, f)
    if info:
        metadata = {
            "iteration": iteration,
            "num_targets": len(df_updated),
            "num_known": int((df_updated[config.main_type_column] != config.unknown_type).sum()),
            "num_unknown": int((df_updated[config.main_type_column] == config.unknown_type).sum()),
        }
        metadata_file = run_path / f"metadata_it_{iteration:02d}.json"

        if not file_exists(metadata_file, verbose):
            with open(metadata_file, "w") as f:
                json.dump(metadata, f, indent=4)
    if verbose:
        print(f"Saved iteration {iteration}")

def finalize_run(run_path, final_df, iteration_history, config, verbose=1):
    final_dir = run_path / "final_results"
    final_dir.mkdir(exist_ok=True)

    print("\n==============================")
    print(" Binary Classification Complete")
    print("==============================\n")
    print("Final galaxy counts:")

    counts = (final_df[config.main_type_column].value_counts())

    for k,v in counts.items():
        print(f"{get_type_name(k,config):15s}: {v:,}")

    final_df.to_csv(final_dir/"final_catalog.csv", index=False)

    print("\nIteration additions:")
    total_added = 0

    for item in iteration_history:
        n = len(item["added"])
        total_added += n
        print(f"Iteration {item['iteration']}: {n:,}")

    print(f"\nTotal added: {total_added:,}")
    print("\nClassifier performance:")
    
    for item in iteration_history:  
        for cid, result in item["results"].items():    
            metrics = result.metrics   
            if "accuracy" in metrics:
                accuracy = metrics["accuracy"]    
            elif "report" in metrics and "accuracy" in metrics["report"]:
                accuracy = metrics["report"]["accuracy"]  
            else:
                accuracy = np.nan   
            print(f"Iteration {item['iteration']} {result.class_name:12s}: {accuracy:.3f}")
            
    print("\nResults saved to:")
    print(final_dir)
```

### The Binary Classifier
    The binary classifier classifies galaxy by asking "Is this a {run_type} galaxy?" 
    It determines yes or no for all galaxies in {data} based on {anchors}.
    
    INPUTS
    ---------------------------------------------------
    data - dtype: string. Path to data to be classified.
    
    anchors - dtype: string. Path to classification anchors.
    
    name - dtype: string. Unique name to be used in all output files.
    
    run_type - dtype: string. Selects which main type is classified.
        
        "all" - classifies all types consecutively.
        "elliptical" - classifies ellipticals.
        "spiral" - classifies spirals.
        "lenticular" - classifies lenticulars.
        "irregular" - classifies irregulars.
    
    mode - dtype: string. Toggle between manual or automatic reiteration of the classifier. 
    If automatic, overrides run_type as "all".
        
        "manual" - saves raw binary results, no combination.
        "automatic - combines binary results using Binary_Eval.py and runs multiple iterations for a final product.

    it_num - dtype: int. Assigns number of iterations.

    source - dtype: string.

        "csv" - reads from CSV
        "pickle" - reads from pickle

    plots - dtype: bool. Saves plots if True.

    info - dtype: bool. Saves run information if True.

    verbose - dtype: int. Level of information printed during run.

        0 - Informs user of saved final product.
        1 - Informs user of all files saved.
        2 - Prints all information produced during run.
        3 - Prints all information and every step of the run. 

    OUTPUTS
    -----------------------------------------------------
    Directory : "{out_dir}/binary_classifier/{name}"
        Contains (up to): 
        --- iterations (e.g. 01, 02, 03)
            --- {name}_it_{iteration}.pkl
            --- binary_classifications (e.g. {name}_{iter}_irregular.csv)
            --- plots
                --- UMAPs
                --- image_scatters
                --- CMs (confusion matrices)
                --- overlays (of UMAP with new classifications)
            --- info
        --- h5_files (e.g. {name}.h5)
        --- processed_images (e.g. {name}.pt)
            

```python
def binary_classifier(config, name, bridge_run_name=None, run_type="all", mode="automatic",
                       it_num=6, plots=True, info=True, verbose=1):
    run_path = prepare_output_directory(config, name)
    classes = get_classes(run_type, config)
    current_df = None
    iteration_history = []

    # The bridge artifact (merged_artifacts.pkl) is written by the UMAP
    # notebook under BRIDGE_RUN_NAME -- defaults to `name` if this run's
    # outputs and the bridge source share one directory, but pass
    # bridge_run_name explicitly when they don't.
    bridge_run_name = bridge_run_name or name
    bridge_run_path = config.output_dir / "binary_classifier" / bridge_run_name

    for iteration in range(1, it_num + 1):
        iteration_path = run_path / "iterations" / f"{iteration:02d}"
        iteration_path.mkdir(parents=True, exist_ok=True)

        if verbose:
            print(f"\n========== Iteration {iteration} ==========")

        if current_df is None:
            gals, embeddings, umap_embeddings = load_merged_artifacts(bridge_run_path)
            np.save(run_path / "umap_coordinates.npy", umap_embeddings)
            df = build_dataframe(gals, embeddings, umap_embeddings, config)
        else:
            df = current_df

        binary_results = {}
        for class_id in classes:
            results = classify_one_type(class_id, df, config, run_path, iteration, plots, verbose)
            save_output(results, iteration, run_path, name, config, plots, verbose)
            binary_results[class_id] = results
        if mode == "manual":
            break
        new_targets, agreement_stats = combine_results(binary_results, config, return_stats=True)
        if len(new_targets) == 0:
            print("No new classifications. Stopping.")
            break
        if plots:
            iteration_plot_dir = (iteration_path / "plots" / "Iteration_Summary")
            iteration_plot_dir.mkdir(parents=True, exist_ok=True)

            save_agreement_histogram(agreement_stats, iteration_plot_dir / "agreement_histogram.png", verbose)
            save_iteration_overlay(df, new_targets, config, iteration_plot_dir / "added_galaxies_overlay.png", verbose)

        current_df = update_catalog(df, new_targets, config)
        iteration_history.append(
            {
                "iteration": iteration,
                "added": new_targets.copy(),
                "agreement_stats": agreement_stats,
                "results": binary_results,
            }
        )
        save_iteration(current_df, iteration, iteration_path, config, info, verbose)

    finalize_run(run_path, current_df, iteration_history, config, verbose)
    return current_df
```

```python
def selectSizeBin(df, binNum, delim=None):
    cutout_size = 152
    pixel_value = 0.262 # arcseconds
    if not delim:
        delim = cutout_size * pixel_value / 60
    df = df[(df['D26'] > binNum*delim) & (df['D26'] <= (binNum+1)*delim)]
    print(f"Selected {len(df)} galaxies less than {(binNum+1)*delim*60} arcseconds")
    return df

selectSizeBin(small_master_update
print(small_master_update['Morphology'].value_counts())
```

```python
supplement = pd.read_csv("/global/cfs/cdirs/desicollab/users/qshimp/anchors/catalog_supplement_0-40.csv")
supplement_filtered = supplement[~supplement['SGAID'].isin(small_master_update['SGAID'])]
supplement_filtered = supplement_filtered.drop(columns = ['timestamp', 'notes', 'status', 'morph_reviewed'])
supplement_filtered
```

```python
# 1. Separate your existing data into what needs to be kept or downsampled
# Elliptical has 1012, so we randomly sample 1000 from it
ellipticals = small_master_update[small_master_update['Morphology'] == 'Elliptical'].sample(n=1000, random_state=42)

# Get the existing counts for the other classes to see how many we need
irreg_existing = small_master_update[small_master_update['Morphology'] == 'Irregular']
spiral_existing = small_master_update[small_master_update['Morphology'] == 'Spiral']
lenti_existing = small_master_update[small_master_update['Morphology'] == 'Lenticular']

# 2. Calculate how many more galaxies are needed for each class
needed_irreg = 1000 - len(irreg_existing)   # 1000 - 996 = 4
needed_spiral = 1000 - len(spiral_existing) # 1000 - 992 = 8
needed_lenti = 1000 - len(lenti_existing)   # 1000 - 807 = 193

# 3. Sample the exact missing amounts from your filtered supplement dataframe
irreg_supp = supplement_filtered[supplement_filtered['Morphology'] == 'Irregular'].sample(n=needed_irreg, random_state=42)
spiral_supp = supplement_filtered[supplement_filtered['Morphology'] == 'Spiral'].sample(n=needed_spiral, random_state=42)
lenti_supp = supplement_filtered[supplement_filtered['Morphology'] == 'Lenticular'].sample(n=needed_lenti, random_state=42)

# 4. Combine the existing data with the supplement data for each class
irreg_final = pd.concat([irreg_existing, irreg_supp])
spiral_final = pd.concat([spiral_existing, spiral_supp])
lenti_final = pd.concat([lenti_existing, lenti_supp])

# 5. Concatenate all four perfectly sized groups into your final dataframe
balanced_master = pd.concat([ellipticals, irreg_final, spiral_final, lenti_final]).reset_index(drop=True)

# Verify the final counts
print(balanced_master['Morphology'].value_counts())
```

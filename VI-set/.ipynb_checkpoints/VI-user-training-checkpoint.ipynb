```python
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
from time import perf_counter

import pandas as pd
from astropy.table import Table
from astropy.io import fits

import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
# Load anchor set
anchors = Table.read("/pscratch/sd/q/qshimp/SGA2020-data/Anchors/VI_4000_sga152x152_complete.csv")

# Load input/output directories
cutout_dir = "/pscratch/sd/q/qshimp/Cutouts/sga2020/Anchors/"
folders = ["Elliptical", "Spiral", "Lenticular", "Irregular"]
out_dir = "/pscratch/sd/q/qshimp/VI_training"
```

```python
# Store column names
ID_COL = 'ref_id'
RA_COL = 'ra'
DEC_COL = 'dec'
MORPH_COL = 'Morphology'

# Prepare user results table
user_results = pd.DataFrame(
    columns=[
        "username",
        "target_id",
        "true_class",
        "user_class",
        "correct",
        "response_time"
    ]
)

# Add user
USERNAME = 'Quillan'

# Save answer statistics
total_answers = 0
correct_answers = 0

# Convert main type to answers
MAIN_TYPE_COL = "Main_type"

main_type_map = {
    20: "E",
    10: "S",
    0: "L",
    -5: "I"
}

# Save type specific statistics
class_stats = {
    "E": {"correct":0, "total":0},
    "S": {"correct":0, "total":0},
    "L": {"correct":0, "total":0},
    "I": {"correct":0, "total":0}
}

confusion_matrix = pd.DataFrame(
    0,
    index=["E", "L", "S", "I"],
    columns=["E", "L", "S", "I"]
)

current_streak = 0
best_streak = 0

MAIN_TYPE_COL = "Main_type"

anchors["trainer_class"] = [
    {
        20: "E",
        10: "S",
        0: "L",
        -5: "I"
    }[m]
    for m in anchors[MAIN_TYPE_COL]
]
```

```python
# Get path to cutout
def cutout_path(row):

    tid = row[ID_COL]
    # Loop through each folder in ~/Anchors
    for folder in folders:
        pattern = os.path.join(cutout_dir, folder, f"{tid}_grz_152_*.fits")
        matches = glob.glob(pattern)

        if matches:
            return matches[0]

    # If an ID is in none of the folders
    raise FileNotFoundError(f"No cutout found for targetid {tid}")

# Load image from fits
def load_image(path):

    with fits.open(path) as hdul:
        data = hdul[0].data

    if data.ndim == 3 and data.shape[0] == 3:

        img = np.transpose(data, (1,2,0))

        # Stretch RGB image for display
        p1, p99 = np.percentile(img, [1, 99])

        if p99 > p1:
            img = np.clip(img, p1, p99)
            img = (img - p1) / (p99 - p1)

        g_img = img[:,:,0]
        r_img = img[:,:,1]
        z_img = img[:,:,2]

    else:

        img = data

        p1, p99 = np.percentile(img, [1, 99])

        img = np.clip(img, p1, p99)
        img = (img - p1) / (p99 - p1)

        g_img = r_img = z_img = img

    return img, g_img, r_img, z_img

# Display rgb image
def display_image(row):
    path = cutout_path(row)
    img, g_img, r_img, z_img = load_image(path)

    print("1 - Elliptical\n2 - Lenticular\n3 - Spiral\n4 - Irregular")
    print(f"TARGETID: {row[ID_COL]}")
    print(f"RA, DEC : {row[RA_COL]:.5f}, {row[DEC_COL]:.5f}")
    print("-" * 50)

    plt.figure(figsize=(10, 8))
    plt.subplot(2,2,1)
    plt.imshow(img, origin="lower")
    plt.title("RGB (approx)")
    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
# Take user input
def get_answer():   
    choices = {"1": "E", "2": "L", "3": "S", "4": "I", "s": "Skip", "stop": "Stop"}
    
    while True:
        cmd = input("Identify [1=Elliptical, 2=Lenticular, 3=Spiral, 4=Irregular, s=skip, stop=quit] : ").strip().lower()
        
        if cmd in choices:
            return choices[cmd]
        
        print("Invalid choice. Try again.")

# Keep track of accuracy statistics
def update_statistics(true_class, answer, correct):

    global total_answers
    global correct_answers
    global current_streak
    global best_streak
    total_answers += 1

    class_stats[true_class]["total"] += 1

    confusion_matrix.loc[
        true_class,
        answer
    ] += 1

    if correct:

        correct_answers += 1

        class_stats[true_class]["correct"] += 1
        
        current_streak += 1
        best_streak = max(best_streak, current_streak)
    
    else:
        current_streak = 0
# Save user results
def save_result(galaxy, answer, true_class, correct, response_time):

    global user_results

    # Add row with enforced datatypes
    new_row = pd.DataFrame({
        "username": [USERNAME],
        "target_id": [int(galaxy[ID_COL])],
        "true_class": [true_class],
        "user_class": [answer],
        "correct": [bool(correct)],
        "response_time": [round(response_time, 2)]
    })

    user_results = pd.concat([user_results, new_row], ignore_index=True)

# Display feedback screen
def show_feedback(correct, answer, true_class, score, total, response_time):
    clear_output(wait=True)
    
    if correct:
        print("✓ CORRECT!")
    else:
        print("✗ INCORRECT")

    print(f"Your answer: {answer}")
    print(f"Correct answer: {true_class}")
    print(f"Time: {response_time:.1f} s")
    print()
    print(f"Score: {score}/{total} ({100*score/total:.1f}%)")
    print(f"Current streak: {current_streak}\nBest streak: {best_streak}")
    print("Loading next galaxy...")
    sleep(2.5) # seconds
```

```python
# Select user
def add_user():

    global USERNAME
    global SAVE_FILE
    global user_results
    global total_answers
    global correct_answers

    USERNAME = input("Enter user name: ").strip()

    SAVE_FILE = (f"/pscratch/sd/q/qshimp/VI_training/savefiles/{USERNAME}.csv")

    # Load existing user
    if os.path.exists(SAVE_FILE):
        user_results = pd.read_csv(SAVE_FILE)
        user_results = user_results.astype({
            "username": str,
            "target_id": int,
            "true_class": str,
            "user_class": str,
            "correct": bool,
            "response_time": float
        })
        total_answers = len(user_results)
        correct_answers = int(user_results["correct"].sum())
        
        # reset
        for morph in class_stats:
            class_stats[morph]["total"] = 0
            class_stats[morph]["correct"] = 0
        
        confusion_matrix.loc[:, :] = 0
        
        for _, row in user_results.iterrows():
        
            true_class = row["true_class"]
            answer = row["user_class"]
        
            class_stats[true_class]["total"] += 1
        
            if row["correct"]:
                class_stats[true_class]["correct"] += 1
        
            if answer in confusion_matrix.columns:
                confusion_matrix.loc[true_class, answer] += 1
            
    # Create new user
    else:
        user_results = pd.DataFrame(
            columns=[
                "username",
                "target_id",
                "true_class",
                "user_class",
                "correct",
                "response_time"
            ]
        )
        total_answers = 0
        correct_answers = 0
        print(f"Created new user: {USERNAME}")
```

```python
# Randomize order
random_indices = np.random.permutation(len(anchors))
shuffled_anchors = anchors[random_indices]

# Select user
add_user()
    
# Initiate galaxy identification
for galaxy in shuffled_anchors:

    # Show galaxy
    clear_output(wait=True)
    display_image(galaxy)
    # Begin timer
    start = perf_counter() 
    answer = get_answer()
    response_time = perf_counter() - start
    
    # End session
    if answer == "Stop":
        clear_output(wait=True)
        user_results.to_csv(SAVE_FILE, index=False)
        print("Saving session...done!")
        break
    
    true_class = main_type_map[galaxy[MAIN_TYPE_COL]]
    
    if answer == "Skip":
        continue
    else: correct = (answer == true_class) # Save right/wrong boolean

    update_statistics(true_class, answer, correct)
    save_result(galaxy, answer, true_class, correct, response_time)
    
    # Save results to user profile
    if total_answers % 5 == 0:
        user_results.to_csv(SAVE_FILE, index=False)

    # Display score page
    show_feedback(correct, answer, true_class, correct_answers, total_answers, response_time)
    plt.close('all')
```
